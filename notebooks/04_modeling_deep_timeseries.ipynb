```python
try:
    import torch
    import torch.nn as nn
    class Tiny(nn.Module):
        def __init__(self):
            super().__init__()
            self.l = nn.Linear(10,2)
        def forward(self,x):
            return self.l(x)
    t = Tiny()
    torch.save(t.state_dict(), '../models/cnn_lstm_model.pt')
    print('Saved placeholder cnn_lstm_model.pt')
except Exception as e:
    with open('../models/cnn_lstm_model.pt','wb') as f:
        f.write(b'Placeholder - install torch and run training')
    print('Torch not available - saved placeholder file')
```
