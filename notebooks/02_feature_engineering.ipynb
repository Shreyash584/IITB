# 02_feature_engineering

Feature engineering template. Produces `data/features.csv`.

```python
import pandas as pd
from pathlib import Path
DATA_DIR = Path('../data')
feat_out = DATA_DIR / 'features.csv'
print('Will save features to', feat_out)
```

```python
try:
    df = pd.read_csv(DATA_DIR / 'clean_EEG_sample.csv')
    numeric = df.select_dtypes(include=['number']).columns
    features = df[numeric].agg(['mean','std']).T.reset_index(drop=True)
    features.to_csv(feat_out, index=False)
    print('Saved example features')
except Exception as e:
    print('No EEG sample found - create features.csv manually or run notebook 01')
```
