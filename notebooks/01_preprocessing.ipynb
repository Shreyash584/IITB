# 01_preprocessing

Preprocessing template for PSY, EEG, GSR files in `../data/`.

```python
import pandas as pd
from pathlib import Path
DATA_DIR = Path('../data')
print('Data directory ->', DATA_DIR.resolve())
```

```python
psy = DATA_DIR / 'PSY.csv'
eeg = DATA_DIR / 'EEG.csv'
gsr = DATA_DIR / 'GSR.csv'
print('Files exist:', psy.exists(), eeg.exists(), gsr.exists())
```

```python
if eeg.exists():
    df = pd.read_csv(eeg)
    for c in ['TimeStamp','Timestamp','Timestamp RAW','Timestamp CAL']:
        if c in df.columns:
            df[c] = pd.to_datetime(df[c], errors='coerce')
            df = df.sort_values(by=c).reset_index(drop=True)
            df.to_csv(DATA_DIR / 'clean_EEG_sample.csv', index=False)
            print('Saved clean_EEG_sample.csv')
            break
```
