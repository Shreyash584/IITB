# 03_modeling_baseline_classification

Train simple baseline classifier and save to `../models/xgb_model.pkl`.

```python
import pandas as pd
from pathlib import Path
import joblib
from sklearn.ensemble import RandomForestClassifier
DATA_DIR = Path('../data')
MODELS_DIR = Path('../models')
feat = DATA_DIR / 'features.csv'
if feat.exists():
    df = pd.read_csv(feat)
    if 'target' not in df.columns:
        df['target'] = (df.iloc[:,0] > df.iloc[:,0].median()).astype(int)
    X = df.drop(columns=['target'])
    y = df['target']
    clf = RandomForestClassifier(n_estimators=10, random_state=42)
    clf.fit(X.fillna(0), y)
    joblib.dump(clf, MODELS_DIR / 'xgb_model.pkl')
    print('Saved baseline model to models/xgb_model.pkl')
else:
    print('features.csv missing; run notebook 02 to create it')
```
